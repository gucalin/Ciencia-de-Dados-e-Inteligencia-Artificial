### Importação das bilbiotecas necessárias

```python
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
import time
```

## 1. Conjunto de dados

```python
# Carregar Dados CIFAR
from tensorflow.keras.datasets import cifar10

# load dataset
(Xtrain, Ytrain), (Xtest, Ytest) = cifar10.load_data()

print('Dados de treinamento:', Xtrain.shape, Ytrain.shape)
print('Dados de teste:', Xtest.shape, Ytest.shape)
```

```python
# Transformar imagens em vetores
x_train = np.reshape(Xtrain, (Xtrain.shape[0], -1))
x_test = np.reshape(Xtest, (Xtest.shape[0], -1))
print('Dimensão dos dados de entrada:', x_train.shape, x_test.shape)

# Calculo da média e desvio padrao dos de treinamento
xmean = np.mean(x_train, axis=0)
xstd = np.std(x_train, axis=0)
print('Média:', xmean.shape)
print('Desvio padrão:', xstd.shape)

# Normalização das imagens
x_train = (x_train - xmean)/xstd
x_test = (x_test - xmean)/xstd

# Codificação one-hot das saídas
y_train_hot = tf.keras.utils.to_categorical(Ytrain, 10)
y_test_hot = tf.keras.utils.to_categorical(Ytest, 10)

print('Dimensão dos dados de entrada:', x_train.shape, x_test.shape)
print('Maximos e mínimos treinamento:', np.min(x_train), np.mean(x_train), np.max(x_train))
print('Maximos e mínimos teste:', np.min(x_test), np.mean(x_test), np.max(x_test))
print('Dimensão dos dados de saída:', y_train_hot.shape, y_test_hot.shape)
```

```python
fig, axs = plt.subplots(2, 8, figsize=(12, 3))
index = 0
for i in range(2):
    for j in range(8):
        axs[i,j].imshow(Xtrain[index], cmap='gray')
        index += 1
plt.show()
```

## 2. Configuração da RNA

```python
# Importa classes
from tensorflow.keras.models import Sequential
from tensorflow.keras import layers

# Confiura modelo
def build_model(n, input_dim):
  model = Sequential()
  model.add(layers.Input(shape=(input_dim)))
  model.add(layers.Dense(units=n[0], activation='relu'))
  model.add(layers.Dense(units=n[1], activation='relu'))
  model.add(layers.Dense(10, activation='softmax'))
  return model

# Cria modelo
input_dim = (3072,)
units = [256, 128]
rna = build_model(units, input_dim)

# Resumo da RNA
rna.summary()
```

## 3. Compilação e treinamento

```python
# Parâmetros de treinamento
Nepocas = 100
Nlote = 1024
```

### 3.1 SGD

```python
# Cria modelo
input_dim = (3072,)
units = [256, 128]
rna = build_model(units, input_dim)

# Resumo da RNA
rna.summary()
```

```python
# Compilação
opt = tf.keras.optimizers.SGD()
rna.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
hist_SGD = rna.fit(x_train, y_train_hot, epochs=Nepocas, batch_size=Nlote, validation_data=(x_test, y_test_hot), verbose=1)
```

```python
results_SGD = rna.evaluate(x_train, y_train_hot, verbose=1)
resulta_SGD = rna.evaluate(x_test, y_test_hot, verbose=1)
```

## Normalização de lote

```python
# Importa classes
from tensorflow.keras.models import Sequential
from tensorflow.keras import layers

# Confiura modelo
def build_model(n, input_dim):
  model = Sequential()
  model.add(layers.Input(shape=(input_dim)))
  model.add(layers.Dense(units=n[0], activation='linear', use_bias=False))
  model.add(layers.BatchNormalization())
  model.add(layers.Activation('relu'))
  model.add(layers.Dense(units=n[1], activation='linear', use_bias=False))
  model.add(layers.BatchNormalization())
  model.add(layers.Activation('relu'))
  model.add(layers.Dense(10, activation='softmax'))
  return model

# Cria modelo
input_dim = (3072,)
units = [256, 128]
rna = build_model(units, input_dim)

# Resumo da RNA
rna.summary()
```

```python
# Compilação
opt = tf.keras.optimizers.SGD(momentum=0.9, nesterov=True)
rna.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
hist_NB = rna.fit(x_train, y_train_hot, epochs=Nepocas, batch_size=Nlote, validation_data=(x_test, y_test_hot), verbose=1)
```
