### Operações entre tipos diferentes

Python não permite certas operações entre tipos de objetos diferentes.

```python
x = 10
y = "5"

print(x + y)
```

Em outros casos, o python converte automaticamente o tipo de uma das variáveis para o tipo mais geral.

```python
x = 5
y = 2.0

resultado = x / y

print(resultado, type(resultado))
```

Em alguns casos, python faz escolhas contraintuitivas:



```python
x = 'ola'
y = 5

print(x * y)
```
