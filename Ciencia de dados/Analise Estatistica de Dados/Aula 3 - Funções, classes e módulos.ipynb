3) Crie duas funções:

calcular_area_base(raio): recebe o raio do círculo e retorna a área da base do cilindro.
* Fórmula: Área = π × raio²

calcular_volume(raio, altura): usa a função calcular_area_base para calcular a área da base e retorna o volume do cilindro.
* Fórmula: Volume = área da base × altura

Calcule a área de um cilindo com raio 7.72. Use π=3.14.

```python
def calcular_area_base(raio):
  area = 3.14 * raio ** 2
  return area

def calcular_volume(raio, altura):
  area_base = calcular_area_base(raio)
  volume = area_base * altura
  return volume

volume = calcular_volume(7.72, 10)
print(f'O volume é de: {volume}')
```

2) Classe Carrinho de Compras:

Crie uma classe chamada CarrinhoDeCompras que represente um carrinho de compras. A classe deve permitir adicionar itens, remover itens e calcular o total da compra.

```python
class Item:
    def __init__(self, nome, preco, quantidade=1):
        self.nome = nome
        self.preco = preco
        self.quantidade = quantidade

    def calcular_total(self):
        return self.preco * self.quantidade

    def __str__(self):
        return f"{self.nome} - {self.quantidade} x R${self.preco:.2f} = R${self.calcular_total():.2f}"


class CarrinhoDeCompras:
    def __init__(self):
        self.itens = []

    def adicionar_item(self, item):
        self.itens.append(item)

    def remover_item(self, nome):
        self.itens = [item for item in self.itens if item.nome != nome]

    def calcular_total(self):
        return sum(item.calcular_total() for item in self.itens)

    def exibir_itens(self):
        if not self.itens:
            print("O carrinho está vazio.")
        else:
            for item in self.itens:
                print(item)

    def __str__(self):
        return f"Total do carrinho: R${self.calcular_total():.2f}"


# Exemplo de uso
if __name__ == "__main__":
    carrinho = CarrinhoDeCompras()
    carrinho.adicionar_item(Item("Maçã", 2.5, 3))
    carrinho.adicionar_item(Item("Banana", 1.8, 2))
    carrinho.adicionar_item(Item("Leite", 4.5, 1))

    print("Itens no carrinho:")
    carrinho.exibir_itens()

    print("\nRemovendo Banana...")
    carrinho.remover_item("Banana")
    carrinho.exibir_itens()

    print("\n", carrinho)
```

### Simulador de Queda Livre

Imagine que você está desenvolvendo um programa para ajudar estudantes de física a calcular o tempo de queda livre de um objeto, sem resistência do ar.

A equação que governa o movimento é:

$$
t = \sqrt{\frac{2h}{g}}
$$

onde:

* t é o tempo de queda (segundos)
* h é a altura da qual o objeto foi solto (metros)
* g é a aceleração da gravidade (9.81 m/s² na Terra)


**Objetivo:**
1. Criar uma classe Objeto com atributos nome e altura.
2. Criar um método calcular_tempo() que retorna o tempo de queda.
3. Criar uma função map() para calcular o tempo de queda de vários objetos.
4. Criar um filtro filter() para selecionar apenas os objetos que levam mais de 2 segundos para cair.
5. Permitir que o usuário insira os objetos manualmente.
6. Exibir o resultado formatado.

```python
class Objeto:
  def __init__(self, nome, altura):
    self.nome = nome
    self.altura = altura

  def calcular_tempo(self):
    g = 9.81
    t = (2 * self.altura / g) ** 0.5
    return t

def main():
    objetos = []
    num_objetos = int(input("Quantos objetos deseja adicionar? "))

    for _ in range(num_objetos):
        nome = input("Nome do objeto: ")
        altura = float(input("Altura de queda (m): "))
        objetos.append(Objeto(nome, altura))

    tempos_de_queda = list(map(lambda obj: (obj.nome, obj.calcular_tempo()), objetos))
    objetos_filtrados = list(filter(lambda obj: obj.calcular_tempo() > 2, objetos))

    print("\nTempos de Queda:")
    for nome, tempo in tempos_de_queda:
        print(f"{nome}: {tempo:.2f}s")

    print("\nObjetos que levam mais de 2 segundos para cair:")
    print(f"{[obj.nome for obj in objetos_filtrados]}")


if __name__ == "__main__":
    main()
```

### Cálculo de Comprimento de Palavras

Crie um programa que receba uma lista de palavras e determine o comprimento de cada uma delas.

**Objetivo:**
1. Criar uma função contar_letras(palavra) que retorna o número de letras da palavra.
2. Pedir ao usuário para inserir uma lista de palavras separadas por espaço.
3. Usar map() para calcular o comprimento de cada palavra.
4. Usar filter() para selecionar apenas palavras com mais de 5 letras.
5. Exibir a lista original de palavras, o comprimento de cada uma e as palavras filtradas.

```python
def contar_letras(palavra):
    return len(palavra)

palavras = input("Digite uma lista de palavras separadas por espaço: ").split()

comprimento = list(map(contar_letras, palavras))

palavras_filtradas = list(filter(lambda palavra: contar_letras(palavra) > 5, palavras))
print("\nLista original de palavras:", palavras)
print("Comprimento de cada palavra:", comprimento)
print("Palavras com mais de 5 letras:", palavras_filtradas)
```

### Soma dos Dígitos dos Números

Crie um programa que receba uma lista de números e determine a soma dos dígitos de cada número.

**Objetivo:**
1. Criar uma função soma_digitos(n) que retorna a soma dos dígitos de um número.
2. Pedir ao usuário para inserir uma lista de números inteiros separados por espaço.
3. Usar map() para calcular a soma dos dígitos de cada número.
4. Usar filter() para selecionar apenas os números cuja soma dos dígitos seja maior que 10.
5. Exibir a lista original, a soma dos dígitos e os números filtrados.

```python
def soma_digitos(n):
    soma = 0
    while n > 0:
        soma += n % 10
        n //= 10
    return soma


numeros = input("Digite uma lista de números inteiros separados por espaço: ").split()

soma = list(map(soma_digitos, map(int, numeros)))

numeros_filtrados = list(filter(lambda n: soma_digitos(n) > 10, map(int, numeros)))

print("\nLista original de números:", numeros)
print("Soma dos dígitos de cada número:", soma)
print("Números cuja soma dos dígitos é maior que 10:", numeros_filtrados)
```
