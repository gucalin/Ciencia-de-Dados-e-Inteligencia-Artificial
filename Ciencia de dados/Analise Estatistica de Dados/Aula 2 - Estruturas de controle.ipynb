**Exemplo 1.** Uma pessoa deseja criar uma poupança de 12350 reais para uma viagem. Todo mês ela consegue economizar 832 reais. Quantos meses ela irá demorar para juntar o dinheiro necessário?

**Exemplo 2.** Imagine que você está criando um jogo simples em que o computador escolhe um número aleatório entre 1 e 100, e o jogador precisa adivinhar esse número. Cada vez que o jogador faz uma tentativa, o computador fornece uma dica se o número correto é maior ou menor. O jogo continua até que o jogador adivinhe o número correto.

```python
import random

numero_secreto = random.randint(1, 100)
tentativa = 0

while True:
    tentativa += 1
    palpite = int(input("Digite seu palpite: "))

    if palpite < numero_secreto:
        print("O número correto é maior. Tente novamente!")
    elif palpite > numero_secreto:
        print("O número correto é menor. Tente novamente!")
    else:
        print(f"Parabéns! Você acertou o número {numero_secreto} em {tentativa} tentativas.")
        break
```

3. Validação de Entrada do Usuário

Peça ao usuário para digitar um número dentro de um determinado intervalo (p. ex., entre 1 e 10). Use um while loop para continuar solicitando a entrada até que um valor válido seja fornecido. Em outras palavras, enquanto não for digitado um valor dentro desse intervalo, o programa continua solicitando a entrada do usuário.

```python
minimo = 1
maximo = 10

while True:
    try:
        numero = int(input(f"Digite um número entre {minimo} e {maximo}: "))
        if minimo <= numero <= maximo:
            print(f"Você digitou um número válido: {numero}")
            break
        else:
            print(f"Por favor, digite um número dentro do intervalo {minimo}-{maximo}.")
    except ValueError:
        print("Entrada inválida! Digite um número inteiro.")
```

4. Descobrindo o Mínimo Múltiplo Comum (MMC):

Dados dois números, use um while loop para encontrar o Mínimo Múltiplo Comum (MMC) entre eles.

```python
# Solicita os dois números ao usuário
num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))

# O MMC começa no maior dos dois números
mmc = max(num1, num2)

# Loop para encontrar o MMC
while True:
    if mmc % num1 == 0 and mmc % num2 == 0:
        print(f"O MMC de {num1} e {num2} é {mmc}")
        break
    mmc += 1  # Incrementa para verificar o próximo múltiplo
```

```python
z = 7

if z % 2 == 0:
    tipo = "par"
else:
    tipo = "ímpar"


if z > 1:
    primo = True
    for i in range(2, int(z ** 0.5) + 1):
        if z % i == 0:
            primo = False
            break
else:
    primo = False

# Exibe o resultado
if primo:
    print(f"O número {z} é {tipo} e primo.")
else:
    print(f"O número {z} é {tipo} e não é primo.")
```

## Pass e Continue

image.png

Exemplo de uso de **pass**:

```python
def minha_funcao():
    pass  # A função existe, mas ainda não faz nada

for i in range(5):
    pass  # O loop existe, mas não executa nada

if True:
    pass  # O if existe, mas ainda não tem código
```

Exemplo de uso de **continue**:

```python
for i in range(5):
    if i == 2:
        continue  # Pula a iteração quando i == 2
    print(i)
```

Comparação entre break, pass e continue:

```python
# Pass
for i in range(5):
    if i == 3:
        pass  # Não faz nada, apenas mantém a estrutura válida
    print(i)
```

```python
# Break
for i in range(5):
    if i == 3:
        break  # Sai do loop quando i == 3
    print(i)
```

```python
# Continue
for i in range(5):
    if i == 3:
        continue  # Pula apenas essa iteração quando i == 3
    print(i)
```
