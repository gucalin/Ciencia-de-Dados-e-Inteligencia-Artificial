**Objetivo do *notebook***: modelar os preços de venda de um modelo de automóvel <BR>em função de sua 'idade' (= ano atual - ano de fabricação).

Será assumida uma relação linear, da forma: $Preço = f(Idade)$
---



```python
# Importação das bibliotecas necessárias
import numpy as np                    # Cálculos em geral
from matplotlib import pyplot as plt  # Traçado de gráficos
from sklearn.linear_model import LinearRegression  # Modelo de regressão linear
```

```python
# Atribuição dos dados de entrada
Idade = np.array([6,7,8,9,10,11,12])
Preco = np.array([70785,68541,62410,59091,53282,49585,45653])
```

```python
# Análise gráfica inicial
# Diagrama de dispersão (scatter plot) dos dados
plt.scatter(x = Idade,y = Preco)
plt.title("Preço x Idade")
plt.xlabel("Anos de uso")
plt.ylabel("Preço de venda (x R$ 10.000,00)")

# Consolidação do gráfico
plt.show()
```

```python
# Preparação dos dados e treinamento do modelo de regressão linear
X = Idade.reshape(-1,1)   # Arranjo bidimensional: (n_pontos,1)
y = Preco                 # Arranjo unidimensional: comprimento = n_pontos

modelo = LinearRegression()
modelo.fit(X,y)
```

```python
# Exibição dos parâmetros do modelo (forma: y = a*x + b)
a = modelo.coef_        # Coeficiente angular
b = modelo.intercept_   # Coeficiente linear (intercepto)

print("Coeficiente angular (depreciação anual) =",a)
print("Intercepto (preço do veículo novo) =",b)
```

```python
# Cálculo de valores extremos previstos pelo modelo
x_reta = np.array([[Idade.min()],[Idade.max()]])  # Coordenadas mínina e máxima de "x"
y_reta = modelo.predict(x_reta)                   # Valores de "y" correspondentes
```

```python
# Gráficos finais: diagrama de dispersão
plt.scatter(x = Idade,y = Preco)
plt.title("Preço x Idade")
plt.xlabel("Anos de uso")
plt.ylabel("Preço de venda (x R$ 10.000,00)")

# Reta de regressão linear
plt.plot(x_reta,y_reta,color = 'red')

# Consolidação dos gráficos
plt.show()
```
