### Objetivo :  
Entender o conceito de ambiente, política, agente e episódio utilizando a biblioteca Gymnasium.


## Bibliotecas Utilizadas

Para a construção do ambiente, utilizaremos agora a biblioteca gymnasium.

```python
import matplotlib.pyplot as plt
from matplotlib import animation
import gymnasium as gym
```

## Variáveis e configurações

Será utilizado um seed para o environment a fim de se obter uma reprodutibilidade no código aqui descrito. Além disso, será usado um parâmetro de controle para definir se o ambiente será estocástico ou não. Um dicionario mapeando o id de cada ação foi definido.

```python
actions_names  = {0: 'LEFT',1: 'DOWN',2: 'RIGHT',3: 'UP'}
is_slippery = True
seed = 42
```

## Definição do Ambiente

**Regras:**  

O jogo inicia com o jogador na posição [0,0], onde o objetivo é chegar ao ponto [3,3].  

Buracos de gelo são distribuidos pelo grid. 

O jogador deve-se mover até atingir um estado terminal ( cair em um buraco ou chegar em seu objetivo).

Nesse exercício o ambiente poderá ser deterministico ou estocástico. is_slippery = True/False.  

**Recompensas:**

* Atingir o objetivo: +1
* Cair em um buraco : 0
* Demais estados    : 0

**Estados:**

![GIF](https://gymnasium.farama.org/_images/frozen_lake.gif)

```python
env = gym.make('FrozenLake-v1',render_mode='rgb_array',is_slippery=is_slippery)
```

## Verificando Dados intrínsecos ao Problema

Ao instanciar um ambiente é possivel verificar o espaço de ações e o espaço de observações.

```python
num_actions = env.action_space.n
num_states = env.observation_space.n
```

```python
print(f'Numero de Ações: {num_actions}')
print(f'Numero de estados: {num_states}')
```

Pode-se também, observar a tabela de transição de cada estado, fazendo: 

`env.unwrapped.P[state][action]`

```python
env.unwrapped.P[0][0]
```

Obtendo: 

`(probability, next_state, reward, done)`

para ambientes deterministicos. 

$P_{ss'} = \mathbb{P}(S_{t+1} = s' \mid S_t = s, A_t = a)$

```python
for state in range(num_states):
    for actions in range(num_actions):
        unwrapped = env.unwrapped.P[state][actions] 
        print(f'Estado {state}, Ação {actions_names[actions]}')
        for num, probs in enumerate(range(len(unwrapped))):
            print(f'P({unwrapped[probs][1]}|{state},{actions_names[actions]}): {unwrapped[probs][0]*100}%, Recompensa:{unwrapped[probs][2]}')
        print('------------------------------------')
```

## Exportando o ambiente como um gif

Utilizaremos todos os frames salvos para a confecção de um gif.

```python
def save_frames_as_gif(frames, filename='Aula1B.gif'):

    plt.figure(figsize=(frames[0].shape[1] / 72.0, frames[0].shape[0] / 72.0), dpi=72)

    patch = plt.imshow(frames[0])
    plt.axis('off')

    def animate(i):
        patch.set_data(frames[i])
        plt.title(f'Step: {i}', fontsize=12, pad=20,loc='left')


    anim = animation.FuncAnimation(plt.gcf(), animate, frames = len(frames), interval=200)
    anim.save(filename)
```

## Definição do episódio

No Gymnasium o episódio é truncado com 100 timesteps - para um ambiente 4x4.

```python
def episode(environment,timestep=100):

    frames = []
    state = environment.reset(seed=seed)[0]
    environment.action_space.seed(seed)
    for _ in range(timestep):
        
        action = environment.action_space.sample()
        print(f'Estado: {state} -> Acao : {actions_names[action]}')
        state, _, terminated, truncated,_ = environment.step(action)
        frames.append(environment.render())

        if terminated or truncated:
            print(f'Estado: {state} -> Acao : -----')
            break
    
    return frames
```

## Instanciando os Objetos

Por fim, usaremos os objetos e funções previamente criados para obter as saídas desejadas.

```python
frames = episode(env)

save_frames_as_gif(frames)
```
