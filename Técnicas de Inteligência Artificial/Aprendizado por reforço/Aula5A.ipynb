## Definição do Ambiente

Considere o seguinte ambiente a ser implementado:

- 𝒮 = {0,1, 2, 3, 4, 5}  
- 𝒮⁺ = {6, Obstáculo}  
- 𝒜 = {$a_1$,$a_2$}  

- $\gamma$ = 0.9
- $\alpha$ = 0.2
- $\epsilon$ = 0.7

Em cada episódio o agente começa no estado $S_0$.


```python
class MDP():
    def __init__(self):
      self.States = ['Estado 0', 'Estado 1', 'Estado 2', 'Estado 3',
                      'Estado 4', 'Estado 5', 'Estado 6', 'Obstaculo']

      self.Rewards = [[-5,-10],
                      [-5,-10],
                      [-5,-10],
                      [100,-100],
                      [-15,-15],
                      [100,-100]]

      self.action_space = ['a1', 'a2']
      self.state = 'Estado 0'

    def isTerminal(self):

      if self.state == 'Estado 6' or self.state == 'Obstaculo':
        return (True)
      else:
        return(False)


    def step(self, action):
      if self.state == 'Estado 0' and action == 'a1':
          reward = self.Rewards[0][0]
          self.state = 'Estado 3'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 0' and action == 'a2':
          reward = self.Rewards[0][1]
          self.state = 'Estado 1'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 1' and action == 'a1':
          reward = self.Rewards[1][0]
          self.state = 'Estado 3'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 1' and action == 'a2':
          reward = self.Rewards[1][1]
          self.state = 'Estado 2'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 2' and action == 'a1':
          reward = self.Rewards[2][0]
          self.state = 'Estado 4'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 2' and action == 'a2':
          reward = self.Rewards[2][1]
          self.state = 'Estado 5'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 3' and action == 'a1':
          reward = self.Rewards[3][0]
          self.state = 'Estado 6'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 3' and action == 'a2':
          reward = self.Rewards[3][1]
          self.state = 'Obstaculo'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 4' and action == 'a1':
          reward = self.Rewards[4][0]
          self.state = 'Estado 3'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 4' and action == 'a2':
          reward = self.Rewards[4][1]
          self.state = 'Estado 5'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 5' and action == 'a1':
          reward = self.Rewards[5][0]
          self.state = 'Estado 6'
          return (self.state, reward, self.isTerminal())

      if self.state == 'Estado 5' and action == 'a2':
          reward = self.Rewards[5][1]
          self.state = 'Obstaculo'
          return (self.state, reward, self.isTerminal())

    def reset(self):
        self.state = 'Estado 0'
        return (self.state)
```

## Definição do Agente

```python
class Agent:
    def __init__(self, env, epsilon=0.7, alpha=0.3, gamma=0.9):
        self.env = env

        self.valid_actions = env.action_space

        self.Q = {'Estado 0':{'a1':0.0,'a2':0.0},
                  'Estado 1':{'a1':0.0,'a2':0.0},
                  'Estado 2':{'a1':0.0,'a2':0.0},
                  'Estado 3':{'a1':0.0,'a2':0.0},
                  'Estado 4':{'a1':0.0,'a2':0.0},
                  'Estado 5':{'a1':0.0,'a2':0.0},
                  'Estado 6':{'a1':0.0,'a2':0.0},
                  'Obstaculo':{'a1':0.0,'a2':0.0}}

        self.epsilon = epsilon
        self.alpha = alpha
        self.gamma = gamma

    def maxQ(self, state):
      return max(self.Q[state].values())

    def action(self, state):

        if random.random() > self.epsilon:
            maximumQ = self.maxQ(state)

            action = random.choice([k for k in self.Q[state].keys()
                                    if self.Q[state][k] == maximumQ])
        else:
            action = random.choice(self.valid_actions)
        return action


    def SARSA(self, state, action, reward, next_state, next_action):

        self.Q[state][action] +=\
         self.alpha * (reward + \
        (self.gamma * self.Q[next_state][next_action])- \
                      self.Q[state][action])

    def QLearn(self, state, action, reward, next_state, is_terminal):

        self.Q[state][action] +=\
         self.alpha * (reward + (self.gamma * self.maxQ(next_state))
                                                - self.Q[state][action])
```

## SARSA (on-policy TD control)

### Instanciando as Classes

```python
env = MDP()
```

```python
A   = Agent(env=env,epsilon=0.7, alpha=0.2, gamma=0.9)
```

### Definindo o Número de Episódios

```python
#Pode ser chamado de epocas
num_episodes = 100000
```

### Aplicando o algortimo

**Inicialização:**
- Q(s, a) ← 0, ∀ s ∈ 𝒮, a ∈ 𝒜

**Repetir para cada episódio (até N episódios):**  
1. Inicializar s  
2. Escolher a a partir de s usando a política 𝜖-greedy baseada em Q  
3. Repetir (para cada passo do episódio):  
   a. Executar ação a, observar recompensa r e próximo estado s'  
   b. Escolher a' a partir de s' usando a política 𝜖-greedy baseada em Q  
   c. Atualizar Q(s, a):  
      → Q(s, a) ← Q(s, a) + α × [r + γ × Q(s', a') − Q(s, a)]  
   d. s ← s'  
   e. a ← a'  
   f. Se s for terminal, sair do loop

**Saída:**  
- Função de ação-valor Q(s, a) aproximada para a política aprendida


```python
state = env.reset()
for episode in range(num_episodes):

  is_terminal = False
  # Escolhendo a ação inicial
  action = A.action(state)
  while is_terminal == False:
      # Realizando a ação
      (next_state, reward, is_terminal)  = env.step(action)

      # Escolhendo a próxima ação
      next_action = A.action(next_state)

      # Aprendendo
      A.SARSA(state, action, reward, next_state, next_action)

      # Incrementando  state and action
      state = next_state
      action = next_action

  #Resetando o environment
  state = env.reset()

  # Fazendo decay dos Hiperparâmetros
  A.epsilon -= 0.6995/num_episodes
  A.alpha -= 0.1995/num_episodes
```
